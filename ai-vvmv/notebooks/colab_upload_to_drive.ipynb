# Colab: Upload files to My Drive
Sử dụng notebook này để: mount Google Drive, upload file từ máy của bạn vào Colab VM, rồi copy vào `MyDrive`.

```python
# 1) Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

# 2) Upload file(s) from your local machine to the Colab VM, then copy to MyDrive
- Run the cell below và dùng nút chọn file để upload.
- Files sẽ được lưu vào thư mục `/content/drive/MyDrive/colab-uploads/` trên Drive.

```python
from google.colab import files
import os

uploaded = files.upload()  # UI để chọn file từ máy bạn
dest_dir = '/content/drive/MyDrive/colab-uploads/'
os.makedirs(dest_dir, exist_ok=True)
for fname, data in uploaded.items():
    path = os.path.join(dest_dir, fname)
    with open(path, 'wb') as f:
        f.write(data)
    print('Saved to', path)
```
